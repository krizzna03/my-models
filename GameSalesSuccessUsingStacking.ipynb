```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
```

```python
df = df.dropna(subset=['total_sales'])
df['success'] = df['total_sales'].apply(lambda x:1 if x>=1.0 else 0)
df['critic_missing'] = df['critic_score'].isnull().astype(int)
```

```python
iter = IterativeImputer(random_state = 42)
num_cols = ['critic_score', 'jp_sales', 'na_sales', 'pal_sales', 'other_sales']
df[num_cols] = iter.fit_transform(df[num_cols])
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

```python
cat_cols = ['console', 'genre']
num_cols = ['critic_score', 'na_sales', 'jp_sales', 'pal_sales', 'other_sales', 'critic_missing']
preprocess = ColumnTransformer([
    ('cat', OneHotEncoder(), cat_cols),
    ('num', StandardScaler(), num_cols)
])
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.ensemble import StackingClassifier
```

```python
base_models = [
    ('lr', LogisticRegression(max_iter=1000)),
    ('rf', RandomForestClassifier(n_estimators=100, random_state=42)),
    ('gb', GradientBoostingClassifier(n_estimators=100, random_state=42)),
    ('svc', SVC(probability=True, kernel='rbf'))
]
meta_model = RandomForestClassifier(n_estimators = 100, random_state = 42)
stack_model = StackingClassifier(
    estimators = base_models,
    final_estimator= meta_model,
    cv = 5
)
```

```python
from sklearn.pipeline import Pipeline
pipeline = Pipeline(steps = [
    ('preprocess', preprocess),
    ('model', stack_model)

])
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score
x = df[num_cols + cat_cols]
y = df['success']
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
pipeline.fit(x_train, y_train)
```

```python
y_pred = pipeline.predict(x_test)
print(classification_report(y_test, y_pred))
print(accuracy_score(y_test, y_pred))
```
