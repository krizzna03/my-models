```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Student Mental health.csv")
df.head()
```

```python
df.info()
```

```python
median = df['Age'].median()
df['Age'].fillna(median, inplace = True)
```

```python
df_copy = df.copy()
for col in ['Do you have Anxiety?', 'Do you have Panic attack?', 'Do you have Depression?']:
    df_copy[col] = df_copy[col].map({'Yes': 1, 'No': 0})
df_copy['mental_issue'] = df_copy[['Do you have Anxiety?', 'Do you have Panic attack?', 'Do you have Depression?']].sum(axis = 1)
def classify(count):
  if count == 0:
    return 1
  elif count > 0 and count <= 2:
    return 2
  else:
    return 3

df_copy['mental_issue'] = df_copy['mental_issue'].apply(classify)
df_copy.head()
```

```python
df = df_copy
df
```

```python
list(df)
```

```python
df.rename(columns={
    'Choose your gender': 'gender',
    'Age': 'age',
    'What is your course?': 'course',
    'Your current year of Study': 'study_year',
    'What is your CGPA?': 'cgpa',
    'Marital status': 'marital_status',
    'Do you have Depression?': 'depression',
    'Do you have Anxiety?': 'anxiety',
    'Do you have Panic attack?': 'panic_attack',
    'Did you seek any specialist for a treatment?': 'sought_treatment',
    'mental_issue': 'mental_issue'  # already fine
}, inplace=True)
```

```python
df.head()
```

```python
copy = pd.get_dummies(df, columns=['course'], prefix='course').copy()
copy
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
copy['gender'] = le.fit_transform(copy['gender'])
copy['marital_status'] = le.fit_transform(copy['marital_status'])
copy['sought_treatment'] = le.fit_transform(copy['sought_treatment'])
copy.head(2)
```

```python
copy['study_year'] = copy['study_year'].str.lower()
copy['study_year'] = copy['study_year'].str.extract('(\d)').astype(int)
```

```python
copy.head(2)
```

```python
df['cgpa'].value_counts()
```

```python
cgpa_map = {
    '0 - 1.99': 1,
    '2.00 - 2.49': 1,
    '2.50 - 2.99': 2,
    '3.00 - 3.49': 2,
    '3.50 - 4.00': 3
}

copy['cgpa'] = copy['cgpa'].map(cgpa_map)
```

```python
copy.fillna(0, inplace = True)
missing = copy.isnull().sum()
missing = missing[missing > 0].sort_values(ascending = False)
missing
```

```python
from sklearn.model_selection import train_test_split, RandomizedSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
x = copy.drop(columns = ['Timestamp', 'mental_issue'])
y = copy['mental_issue']
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.3, stratify = y, random_state = 42)
print(y.value_counts())
```

```python
log_model = LogisticRegression(max_iter = 1000, solver = 'lbfgs', class_weight = 'balanced')
log_model.fit(x_train, y_train)
log_pred = log_model.predict(x_test)
log_accuracy = accuracy_score(y_test, log_pred)
print(f"Logistic Regression Accuracy: {log_accuracy}")
print(classification_report(y_test, log_pred))
```

```python
from sklearn.model_selection import cross_val_score
cross_val_score(log_model, x, y, cv=5, scoring='f1_weighted')
```

```python
print(np.array_equal(x_train, x_test))
print(copy.duplicated().sum())
print(x_train.shape, x_test.shape)
```

```python
forest = RandomForestClassifier(n_estimators = 100, random_state = 42, class_weight = 'balanced')
forest.fit(x_train, y_train)
forest_pred = forest.predict(x_test)
forest_accuracy = accuracy_score(y_test, forest_pred)
print(f"Random Forest Accuracy: {forest_accuracy}")
print(classification_report(y_test, forest_pred))
```

```python
from sklearn.multiclass import OneVsRestClassifier
from sklearn.svm import SVC

model = OneVsRestClassifier(SVC(kernel='linear'))
model.fit(x_train, y_train)
svm_pred = model.predict(x_test)
svm_accuracy = accuracy_score(y_test, svm_pred)
print(f"SVM Accuracy: {svm_accuracy}")
print(classification_report(y_test, svm_pred))
```
